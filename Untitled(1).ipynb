```python
from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments
from datasets import Dataset
import torch

FIM_SEP = "<FIM-SEP>"

# Load tokenizer and model
model_name = "gpt2"  # replace with your model (e.g. CodeLLaMA, Qwen2.5)
tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.add_tokens([FIM_SEP])  # add the separator token if not present

model = AutoModelForCausalLM.from_pretrained(model_name)
model.resize_token_embeddings(len(tokenizer))

# Example data
data = [
    {
        "middle": "c = Report",
        "ground_truth": "(\"bob\", 21, )"
    }
]
dataset = Dataset.from_list(data)

# Preprocessing function
def preprocess(example):
    input_text = example["middle"] + " " + FIM_SEP + " " + example["ground_truth"]
    input_ids = tokenizer(input_text, return_tensors="pt", padding=False).input_ids[0]

    sep_index = input_text.index(FIM_SEP) + len(FIM_SEP)
    target_start = len(tokenizer(input_text[:sep_index], return_tensors="pt").input_ids[0])

    labels = torch.full_like(input_ids, -100)
    labels[target_start:] = input_ids[target_start:]

    return {
        "input_ids": input_ids,
        "labels": labels,
        "attention_mask": torch.ones_like(input_ids)
    }

tokenized_dataset = dataset.map(preprocess)

# Padding collator
def collate(batch):
    return tokenizer.pad(batch, return_tensors="pt")

# Training arguments
training_args = TrainingArguments(
    output_dir="./fim-fill",
    per_device_train_batch_size=2,
    bf16=true
    learning_rate=5e-5,
    num_train_epochs=3,
    logging_steps=10,
    save_steps=100,
    warmup_ratio=0.5,
    
)

# Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    tokenizer=tokenizer,
    data_collator=collate
)

trainer.train()
```
