```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "PySage"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
from datasets import load_dataset
import numpy as np

def load_training_data(data_path):
    dataset = load_dataset("json", data_files=data_path, split="train")
    return dataset
    
def tokenize_function(example, tokenizer):
    tokens = example["tokens"]
    out_dict = {
        "input_ids": tokens,
        "labels": tokens,
        "attention_mask": [1] * len(tokens)
    }
    return {k: np.array(v) for k, v in out_dict.items()}

train_dataset_path = "python-dataset/new_train_data.jsonl"
val_dataset_path = "python-dataset/new_val_data.jsonl"
dataset = load_training_data(train_dataset_path)
train_dataset = dataset.map(lambda x: tokenize_function(x, tokenizer), batched=True)
dataset = load_training_data(val_dataset_path)
val_dataset = dataset.map(lambda x: tokenize_function(x, tokenizer), batched=True )
```

```python
import json
import torch
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments





def train_model(output_dir, batch_size=4, epochs=1, lr=9e-5):


    training_args = TrainingArguments(
        output_dir=output_dir,
        per_device_train_batch_size=batch_size,
        per_device_eval_batch_size=batch_size,
        gradient_accumulation_steps=128,
        bf16=True,
        num_train_epochs=epochs,
        learning_rate=lr,
        weight_decay=0.01,
        save_total_limit=10,
        eval_steps=100,
        save_strategy="steps",
        eval_strategy="steps",
        save_steps = 500,
        logging_dir="./logs",
        logging_steps=10,
        logging_strategy="steps",
        logging_first_step=True,
        report_to=["tensorboard"],
        push_to_hub=False,
        lr_scheduler_type="cosine",
        warmup_ratio = 0.05, 
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
    )

    try:
        trainer.train()
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        print("CUDA out of memory! Cleared cache. Try reducing batch size or sequence length.")

    model.save_pretrained(output_dir)
    tokenizer.save_pretrained(output_dir)
    print(f"Model saved to {output_dir}")


if __name__ == "__main__":
    
    model_name = "PySage"
    output_dir = "trained"

    train_model(output_dir)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Configuration
total_steps = 750
warmup_steps = int(0.05 * total_steps)  # 5% of total steps
final_lr = 1e-4

# Create learning rate schedule
lr_schedule = np.ones(total_steps) * final_lr
lr_schedule[:warmup_steps] = np.linspace(0, final_lr, warmup_steps)

# Plotting
plt.figure(figsize=(10, 4))
plt.plot(np.arange(total_steps), lr_schedule, label="Logging Rate", color="green", linewidth=2)
plt.axvline(warmup_steps, color='gray', linestyle='--', label='Warmup End (~5%)')
plt.title("Learning Rate Schedule with Warmup")
plt.xlabel("Training Steps")
plt.ylabel("Learning Rate")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
